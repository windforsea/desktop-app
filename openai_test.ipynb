# OpenAI Responses API 기본 요청 및 응답 테스트

> **참고**: 최신 OpenAI 권장 인터페이스인 **Responses API** (`client.responses.create`)를 사용합니다.
> - 기존 `chat.completions.create` 대신 단일 진입점인 `client.responses.create`를 사용합니다.
> - 시스템 지침은 상위 레벨 파라미터 `instructions`로 전달합니다.
> - 사용자 입력은 `input` 파라미터에 문자열 또는 메시지 리스트 형태로 전달합니다.
> - 결과 텍스트는 `response.output_text` 헬퍼로 간편하게 추출할 수 있습니다.

```python
import os
from dotenv import load_dotenv
from openai import OpenAI

# 1. .env 환경 변수 로드 (OPENAI_API_KEY)
load_dotenv()

# 2. OpenAI 클라이언트 초기화
# 환경 변수 OPENAI_API_KEY를 자동 참조합니다. 직접 전달 시: OpenAI(api_key="...")
client = OpenAI()

# 3. 최신 Responses API 요청
response = client.responses.create(
    model="gpt-5.6-luna",
    instructions="친절하고 명확하게 답변하는 한국어 어시스턴트입니다.",
    input="안녕하세요! 간단하게 한 줄로 자기소개해 주세요."
)

# 4. 응답 텍스트 출력 (최신 Responses SDK 전용 output_text 헬퍼)
print(response.output_text)
```
